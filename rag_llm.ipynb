# RAG + LLM
Installare il pacchetto, preparare i documenti e avviare Ollama.

```python
from ragmens_core import RagConfig, RAGPipeline
config = RagConfig.from_env()
rag = RAGPipeline.prepare(config)
```

```python
domanda = "Quali sono le conclusioni principali?"
preview = rag.preview(domanda)
print(preview["context"])
```

```python
risultato = rag.ask(domanda)
print(risultato["answer"])
print(risultato["sources"])
```

```python
rag.close()
```
